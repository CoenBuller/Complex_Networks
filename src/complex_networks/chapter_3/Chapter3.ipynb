```python
#write your own Python code to generate a random network with given number of nodes and edge probabilities using NetworkX function
import networkx as nx
import matplotlib.pyplot as plt
import random

# the predicted number of edges: p x N x (N - 1) / 2
#def predicted_number_of_edges(p, N):
#	return (p * N * (N - 1)) / 2

# number of nodes
n = 12
# probability p
p = 0.5

# our graph
G = nx.Graph()

# start with adding n nodes
for i in range (0, n):
	G.add_node(i)

# consider every possible link between each node
for i in range (0, n):
	for j in range (i, n):
		if (random.random() < p or p == 1.0) & (i != j):
			G.add_edge(i, j)

print("Actual number of edges: " + str(G.number_of_edges()))

nx.draw(G, pos=nx.circular_layout(G))
plt.show()
```

```python
#create random networks of different size and probabilities and compare their properties.

import numpy as np
import matplotlib.pyplot as plt
import networkx as nx

n1 = 10
n2 = 50

er1 = nx.erdos_renyi_graph(n1, 0.5/n1, directed=False)
er2 = nx.erdos_renyi_graph(n1, 1/n1, directed=False)
er3 = nx.erdos_renyi_graph(n1, np.log(n1)/n1, directed=False)
er4 = nx.erdos_renyi_graph(n2, 0.5/n2, directed=False)
er5 = nx.erdos_renyi_graph(n2, 1/n2, directed=False)
er6 = nx.erdos_renyi_graph(n2, np.log(n2)/n2, directed=False)

fig, axs = plt.subplots(2, 3, figsize=(15, 10))
axs[0, 0].set_title("n=10, np=0.5")
nx.draw(er1, ax=axs[0, 0], pos=nx.spring_layout(er1), with_labels=False)
axs[0, 1].set_title("n=10, np=1")
nx.draw(er2, ax=axs[0, 1], pos=nx.spring_layout(er2), with_labels=False)
axs[0, 2].set_title("n=10, np=log(n)")
nx.draw(er3, ax=axs[0, 2], pos=nx.spring_layout(er3), with_labels=False)
axs[1, 0].set_title("n=50, np=0.5")
nx.draw(er4, ax=axs[1, 0], pos=nx.spring_layout(er4), with_labels=False)
axs[1, 1].set_title("n=50, np=1")
nx.draw(er5, ax=axs[1, 1], pos=nx.spring_layout(er5), with_labels=False)
axs[1, 2].set_title("n=50, np=log(n)")
nx.draw(er6, ax=axs[1, 2], pos=nx.spring_layout(er6), with_labels=False)

plt.tight_layout()
plt.show()

```

```python
#code to plot size of the largest connected component

import numpy as np
import matplotlib.pyplot as plt
import networkx as nx

gcc = []  # Initialize gcc, probability
probability = []  # to empty lists

for i in range(1, 71):
    prob = 0.0002 * i
    probability.append(prob)  # Append probability
    er = nx.erdos_renyi_graph(500, prob)  # Generate Erdos-Renyi graph
    comp = nx.connected_components(er)  # Compute list of clusters
    largest = max(len(c) for c in comp)  # max over cluster sizes
    gcc.append(largest)  # Append size of largest cluster

plt.scatter(np.array(probability) * 500, gcc, color='blue')  # Use a color for points
plt.xlabel("np")
plt.ylabel("Number of vertices")
plt.title("Size of the largest connected component")
plt.show()
```

```python
#code to plot average path length for random networks

import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Function to compute the average path length of a random graph
def average_path_length(graph_size, edge_probability):
    G = nx.erdos_renyi_graph(graph_size, edge_probability)
    if nx.is_connected(G):
        return nx.average_shortest_path_length(G)
    else:
        # For disconnected graphs, return None or handle appropriately
        return None

# Parameters
sizes = range(10, 101, 10)  # Different sizes for the random graphs
edge_probability = 0.1      # Probability of edge creation

# Compute the average path length for each graph size
average_path_lengths = []
for size in sizes:
    lengths = [average_path_length(size, edge_probability) for _ in range(10)]
    lengths = [length for length in lengths if length is not None]  # Filter out None values
    if lengths:
        average_path_lengths.append(np.mean(lengths))
    else:
        average_path_lengths.append(None)

# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(sizes, average_path_lengths, marker='o')
plt.xlabel('Graph Size (number of nodes)')
plt.ylabel('Average Path Length')
plt.title('Average Path Length for Random Graphs of Different Sizes (p=0.1)')
plt.grid(True)
plt.show()
```

```python
#generate random and BA network. Compare their degree distribution

import numpy as np
import matplotlib.pyplot as plt
import networkx as nx

# Generate Erdos-Renyi graph
er = nx.erdos_renyi_graph(10000, 0.001, directed=False)
# Generate Barabasi-Albert graph
br = nx.barabasi_albert_graph(10000, 1)

# Degree distribution for Erdos-Renyi graph
d_er = np.array(list(dict(er.degree()).values()))
degree_counts_er = np.bincount(d_er)
p_er = degree_counts_er / np.sum(degree_counts_er)

# Plotting degree distribution for Erdos-Renyi graph
plt.figure()
plt.loglog(np.nonzero(p_er)[0], p_er[np.nonzero(p_er)], 'ro', label='Erdos-Renyi')
plt.xlabel('Degree d')
plt.ylabel('p(d)')
plt.title('Degree distribution in log-log')
plt.grid(True)

# Degree distribution for Barabasi-Albert graph
d_br = np.array(list(dict(br.degree()).values()))
degree_counts_br = np.bincount(d_br)
p_br = degree_counts_br / np.sum(degree_counts_br)

# Plotting degree distribution for Barabasi-Albert graph
plt.loglog(np.nonzero(p_br)[0], p_br[np.nonzero(p_br)], 'ko', label='Barabasi-Albert')
plt.legend()
plt.show()

```

```python
#code to generate adjacency matrix for a random network with given number of nodes and edge connection probabilities.

import numpy as np

def ergraphp(n, p):
    # create a matrix of zeros with n rows and n columns
    m = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            # if source (i) and destiny (j) are different,
            # create edge with probability p.
            if i != j and np.random.rand() <= p:
                m[i, j] = 1
    return m

n=10
p=0.1
print(ergraphp(n, p))
```

```python
#code to create Disc network

import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

def discgraph(n, r, showgraph=0):
    centers = np.random.rand(n, 2)
    m = np.zeros((n, n))

    for i in range(n):
        for j in range(n):
            if i != j:
                dist = np.sqrt((centers[i, 0] - centers[j, 0]) ** 2 +
                               (centers[i, 1] - centers[j, 1]) ** 2)
                if dist < 2 * r:
                    m[i, j] = 1

    if showgraph:
        g = nx.from_numpy_array(m)
        pos = {i: centers[i] for i in range(n)}
        nx.draw(g, pos, with_labels=True, arrows=False)
        plt.show()

    return m

discgraph(200, 0.1, showgraph=1)
```
