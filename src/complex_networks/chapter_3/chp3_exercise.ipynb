```python


def random_graph(n_nodes, p):

    node_list = range(1, n_nodes+1)

    G = nx.Graph()
    G.add_nodes_from(node_list)

    for pair in combinations(node_list, 2):
        if np.random.rand() < p:
            G.add_edge(pair[0], pair[1])
    return G

n = 100
l=1.1
G = random_graph(n_nodes=n, p=l/n)
nx.draw(G)
plt.show()
```

```python
def centrality_measures(graph: nx.Graph, measure: str='closeness', **kwargs):
    measures = {
        "closeness": nx.closeness_centrality,
        "degree": nx.degree_centrality,
        "betweenness": nx.betweenness_centrality,
        "pagerank": nx.pagerank,
        "core_number": nx.core_number,
        }

    if measure not in measures:
        raise ValueError(f"The measure is not one of the valid measures. Choose one of: {measures.keys()}")

    f = measures[measure]
    m = f(graph, **kwargs)

    return m
```
